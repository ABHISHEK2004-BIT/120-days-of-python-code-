```python
a =int(input("Enter the First Number="))
b =int(input("Enter the Second Number="))

ans1=a+b
print("The addition of two number:",ans1)

ans2=a-b
print("The Subtraction of two number:",ans2)

ans3=a*b
print("The multiplication of two number:",ans3)

ans4=a/b
print("The division of two number:",ans4)

ans5=a//b
print("The floor division of two number:",ans5)

ans6=a%b
print("The remender of two number:",ans6)

ans7=a**b
print("The exponentiation of two number:",ans7)
```
